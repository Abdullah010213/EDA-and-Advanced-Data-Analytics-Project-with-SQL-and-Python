```python
product_sales = sales.merge(
    products,
    on='product_key',
    how='left',
    )

product_sales['order_year'] = product_sales['order_date'].dt.year

Yearly_Product_Sales = product_sales.groupby(['order_year', 'product_name']).agg(
    current_sales = ('sales_amount', 'sum')
).sort_values(['product_name', 'order_year']).dropna()

Yearly_Product_Sales['average_sales'] = Yearly_Product_Sales.groupby('product_name')['current_sales'].transform('mean')

Yearly_Product_Sales['diff_average'] = Yearly_Product_Sales['current_sales'] - Yearly_Product_Sales['average_sales']

Yearly_Product_Sales['average_status'] = np.where(Yearly_Product_Sales['diff_average'] > 0, 'Above Average',
                                            np.where(Yearly_Product_Sales['diff_average'] < 0, 'Belove Average', 'Average'))

Yearly_Product_Sales['previouos_Sales'] = Yearly_Product_Sales.groupby('product_name')['current_sales'].shift(1)

Yearly_Product_Sales['diff_previouos_Sales'] = Yearly_Product_Sales['current_sales'] - Yearly_Product_Sales['previouos_Sales']

Yearly_Product_Sales['diff_previouos_Sales_status'] = np.select([Yearly_Product_Sales['diff_previouos_Sales'] > 0, Yearly_Product_Sales['diff_previouos_Sales'] < 0],
                                                                ['Increase', 'Decrease'],
                                                                default='No Change')

Yearly_Product_Sales = Yearly_Product_Sales.sort_values(['product_name', 'order_year'])

Yearly_Product_Sales.to_csv('Yearly_Product_Sales_Performance')

Yearly_Product_Sales
```

```python
Customer_Sales = sales.merge(
    customers,
    on='customer_key',
    how='left'
)

Customer_Spending = Customer_Sales[['customer_key']].copy()

Customer_Spending = Customer_Sales.groupby('customer_key').agg(
    total_spending = ('sales_amount', 'sum'),
    first_order = ('order_date', 'min'),
    last_order = ('order_date', 'max')    
).reset_index()

Customer_Spending['lifespan'] = (
    (Customer_Spending['last_order'].dt.year - Customer_Spending['first_order'].dt.year) * 12
    + (Customer_Spending['last_order'].dt.month - Customer_Spending['first_order'].dt.month)
    - (Customer_Spending['last_order'].dt.day < Customer_Spending['first_order'].dt.day).astype(int)
)

Customer_Spending['customer_segment'] = np.select(
                                        [(Customer_Spending['lifespan'] >= 12) & (Customer_Spending['total_spending'] > 5000), 
                                         (Customer_Spending['lifespan'] >= 12) & (Customer_Spending['total_spending'] <= 5000)],
                                        ['VIP', 'Regular'],
                                        default='New'
)

Customer_Spending_by_Grouping = Customer_Spending.groupby('customer_segment')['customer_key'].count().reset_index(name='total_customers')

Customer_Spending_by_Grouping.set_index('customer_segment')['total_customers'].plot(
    kind='pie',
    autopct='%1.1f%%',
    ylabel=''
)
plt.title('Total Customer by Customer Segment')
```

```python
'''============================================================
Customer Report
============================================================
Purpose:
- This report consolidates key customer metrics and behaviors

Highlights:
1. Gathers essential fields such as names, ages, and transaction details.
2. Segments customers into categories (VIP, Regular, New) and age groups.
3. Aggregates customer-level metrics:
   - total orders
   - total sales
   - total quantity purchased
   - total products
   - lifespan (in months)
4. Calculates valuable KPIs:
   - recency (months since last order)
   - average order value
   - average monthly spend
============================================================'''

customer_sales_total = sales.merge(customers, on='customer_key', how='left')

customer_sales = customer_sales_total[['order_number', 'product_key', 'order_date', 'sales_amount', 'quantity', 'customer_key', 'customer_number']]

customer_sales['customer_name'] = customer_sales_total['first_name'] + " " + customer_sales_total['last_name']
customer_sales['age'] = (date.today().year - customer_sales_total['birthdate'].dt.year).astype('Int64')   


customer_aggregation = customer_sales.groupby(['customer_key', 'customer_number', 'customer_name', 'age']).agg(
                        total_orders = ('order_number', 'nunique'),
                        total_sales = ('sales_amount', 'sum'),
                        total_quantity = ('quantity', 'sum'),
                        total_product = ('product_key', 'nunique'),
                        first_order_date = ('order_date', 'min'),
                        last_order_date = ('order_date', 'max')
                  ).reset_index()

customer_aggregation['lifespan'] = (
    (customer_aggregation['last_order_date'].dt.year - customer_aggregation['first_order_date'].dt.year) * 12
    + (customer_aggregation['last_order_date'].dt.month - customer_aggregation['first_order_date'].dt.month)
    - (customer_aggregation['last_order_date'].dt.day < customer_aggregation['first_order_date'].dt.day).astype(int)
)

customer_report = customer_aggregation.copy()


customer_report['age_group'] = np.select(
                              [customer_report['age'] < 20, (customer_report['age'] >= 20) & (customer_report['age'] < 30),
                               (customer_report['age'] >= 30) & (customer_report['age'] < 40),
                               (customer_report['age'] >= 40) & (customer_report['age'] < 50)],
                              ['Under 20', '20-29', '30-29', '40-49'],
                              default='50 and Above'
)

customer_report['customer_segment'] = np.select(
                                       [(customer_report['lifespan'] >= 12) & (customer_report['total_sales'] > 5000),
                                        (customer_report['lifespan'] >= 12) & (customer_report['total_sales'] <= 5000)],
                                       ['VIP', 'Regular'],
                                       default='New'
)

customer_report['AOV'] = np.where(customer_report['lifespan'] == 0, 
                                  0, 
                                  (customer_report['total_sales'] / customer_report['lifespan']).round(2))

nonzero = customer_report['lifespan'].notna() & customer_report['lifespan'].ne(0)

avg_monthly_spend = (
    customer_report.loc[nonzero, 'total_sales']
    .div(customer_report.loc[nonzero, 'lifespan'])
    .mean()
    .round(2)
)

customer_report['avg_monthly_spend'] = np.where(
    customer_report['lifespan'].eq(0),
    customer_report['total_sales'],
    avg_monthly_spend
)

customer_report.to_csv('Customer_Reprot')

customer_report
```
